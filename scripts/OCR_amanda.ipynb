```python
import os
from mistralai import Mistral

# --- 1. Load API key from environment ---
os.environ["MISTRAL_API_KEY"] = "RIp6OK3YshcjUv2oN1WIVUjHymlz7HIJ"

api_key = os.environ["MISTRAL_API_KEY"]
client = Mistral(api_key=api_key)

# --- 2. PDF path ---
pdf_path = "/Users/amandawhite/Downloads/Bridgeport Series 1 Milling manual Installation, Operation, Maintenance, and Parts list.pdf"

# --- 3. Upload for OCR ---
uploaded_pdf = client.files.upload(
    file={"file_name": pdf_path.split("/")[-1], "content": open(pdf_path, "rb")},
    purpose="ocr",
)

# --- 4. Get signed URL ---
signed_url = client.files.get_signed_url(file_id=uploaded_pdf.id)

# --- 5. OCR process ---
ocr_response = client.ocr.process(
    model="mistral-ocr-latest",
    document={
        "type": "document_url",
        "document_url": signed_url.url
    },
    include_image_base64=True
)
# --- 6. Save result to Markdown file ---
# ocr_response is a Pydantic model with a `.pages` list
pages = getattr(ocr_response, "pages", []) or []

md_parts = []
for p in pages:
    md = getattr(p, "markdown", "") or ""
    if md:
        md_parts.append(md)

full_md = "\n\n".join(md_parts)
with open("bridgeport_manual_ocr.md", "w") as f:
    f.write(full_md)

print(f"✅ OCR text saved to bridgeport_manual_ocr.md ({len(md_parts)} pages with markdown)")

```
